# Pizza Sales Analysis

## 1. Introduction

Pizza sales performance is an important aspect of understanding customer purchasing behavior, product demand, revenue generation, and operational efficiency within a pizza business. Analyzing sales data can help businesses identify their best-performing products, understand customer ordering patterns, optimize inventory, and make informed decisions about pricing, promotions, and product offerings.

This project analyzes pizza sales data to identify patterns and trends in sales performance throughout the year. The analysis examines key business metrics such as total revenue, total pizzas sold, total orders, average order value, and average pizzas per order. It also explores sales performance across pizza categories, pizza sizes, individual pizza products, ingredients, days of the week, hours of the day, and months of the year.

The dataset contains **48,620 sales records and 12 variables**, including order information, pizza quantity, order date and time, unit price, total price, pizza size, category, ingredients, and pizza name.

## key performance indicators:

* **Total Revenue:** $817,860.05
* **Total Pizzas Sold:** 49,574
* **Total Orders:** 21,350
* **Average Order Value:** $38.31
* **Average Pizzas per Order:** 2.32

## Project Objectives

The main objectives of this analysis are to:

* Evaluate the overall sales performance of the pizza business.
* Analyze total revenue, total orders, and total pizzas sold.
* Determine the average order value and average number of pizzas purchased per order.
* Examine sales trends across different days of the week.
* Identify hourly patterns in customer orders and revenue.
* Analyze monthly sales and revenue trends throughout the year.
* Compare sales performance across different pizza categories.
* Examine the relationship between pizza size and sales across categories.
* Identify the most frequently used pizza ingredients.
* Identify the top-performing pizzas based on quantity sold, number of orders, and revenue.
* Identify lower-performing pizzas based on quantity, orders, and revenue.
* Generate insights that can support better inventory, marketing, product, and sales decisions.

## Business Problem

The pizza business wants to better understand its sales performance and customer purchasing patterns. Although sales transactions are being recorded, management needs a clearer understanding of which products and categories generate the most demand and revenue, when customers are most likely to place orders, and which products may require further attention.

The analysis will provide data-driven insights into product performance, customer ordering patterns, revenue trends, pizza sizes, categories, and ingredient demand. These insights can help management make informed decisions regarding inventory planning, promotional activities, staffing during peak periods, product positioning, and the evaluation of underperforming products.

#### Daily Trend-Total Orders

```python
df["order_date"] = pd.to_datetime(df["order_date"])

df["day_name"] = df["order_date"].dt.day_name()

weekday_order = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"]

df["day_name"] = pd.Categorical(df["day_name"], categories=weekday_order, ordered=True)

orders_by_day = df.groupby("day_name", observed=False)["order_id"].nunique()

ax = orders_by_day.plot(kind="bar", figsize=(8, 5), color="Purple", edgecolor="black")

ax.set_title("Total Orders by Day of Week")
ax.set_xlabel("Day of Week")
ax.set_ylabel("Number of Orders")

ax.bar_label(ax.containers[0], fmt="%d")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

#### Daily Trend-Total Revenue

```python
df["order_date"] = pd.to_datetime(df["order_date"])

df["day_name"] = df["order_date"].dt.day_name()

weekday_order = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"]

df["day_name"] = pd.Categorical(df["day_name"], categories=weekday_order, ordered=True)

orders_by_day = df.groupby("day_name", observed=False)["total_price"].sum()

ax = orders_by_day.plot(kind="bar", figsize=(8, 5), color="green", edgecolor="black")

ax.set_title("Total Orders by Day of Week")
ax.set_xlabel("Day of Week")
ax.set_ylabel("Total Revenue")

ax.bar_label(ax.containers[0], fmt="%.2f")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

#### Hourly Trend-Total Orders

```python
df["order_time"] = pd.to_datetime(df["order_time"], format="%H:%M:%S")

df["order_hour"] = df["order_time"].dt.hour

orders_by_hour = df.groupby("order_hour", observed=False)["order_id"].nunique()

ax = orders_by_hour.plot(kind="bar", figsize=(10, 5), color="maroon", edgecolor="black")

ax.set_title("Total Orders by Day")
ax.set_xlabel("Hour of Days(24hr format)")
ax.set_ylabel("Number of Orders")

ax.bar_label(ax.containers[0], fmt="%d")

plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

#### Hourly Trend-Total Revenue

```python
df["order_time"] = pd.to_datetime(df["order_time"], format="%H:%M:%S")

df["order_hour"] = df["order_time"].dt.hour

orders_by_hour = df.groupby("order_hour", observed=False)["total_price"].sum()

ax = orders_by_hour.plot(kind="bar", figsize=(10, 5), color="maroon", edgecolor="black")

ax.set_title("Total Orders by Day")
ax.set_xlabel("Hour of Days(24hr format)")
ax.set_ylabel("Number Revenue")

ax.bar_label(ax.containers[0], fmt="%.2f")

plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

#### Monthly Trend-Total Order

```python
df['order_date'] = pd.to_datetime(df['order_date'], dayfirst=True)

df['month_name'] = df['order_date'].dt.month_name()

month_order = ['January', 'February', 'March', 'April', 'May', 'June',
               'July', 'August', 'September', 'October', 'November', 'December']

df['month_name'] = pd.Categorical(df['month_name'], categories=month_order, ordered=True)

orders_by_month = df.groupby('month_name', observed=True)['order_id'].nunique()

plt.figure(figsize=(8,5))
plt.fill_between(orders_by_month.index, orders_by_month.values, color='orange', alpha=0.3)
plt.plot(orders_by_month.index, orders_by_month.values, color='black', linewidth=2, marker='o')

plt.title("Total Orders by Month")
plt.xlabel("Month")
plt.ylabel("Number of Orders")
plt.xticks(rotation=45)

for i, val in enumerate(orders_by_month):
    plt.text(i, val + 20, f'{int(val)}', ha='center', va='bottom', fontsize=9)

plt.tight_layout()
plt.show()
```

#### Monthly Trend-Total Revenue

```python
df['order_date'] = pd.to_datetime(df['order_date'], dayfirst=True)

df['month_name'] = df['order_date'].dt.month_name()

month_revenue = ['January', 'February', 'March', 'April', 'May', 'June',
               'July', 'August', 'September', 'October', 'November', 'December']

df['month_name'] = pd.Categorical(df['month_name'], categories=month_revenue, ordered=True)

revenue_by_month = df.groupby('month_name', observed=False)['total_price'].sum()

plt.figure(figsize=(10,5))
plt.fill_between(revenue_by_month.index, revenue_by_month.values, color='orange', alpha=0.3)
plt.plot(revenue_by_month.index, revenue_by_month.values, color='black', linewidth=1, marker='o')

plt.title("Total Revenue by Month")
plt.xlabel("Month")
plt.ylabel("Number of Revenue")
plt.xticks(rotation=45)

for i, val in enumerate(revenue_by_month):
    plt.text(i, val + 20, f'{int(val)}', ha='center', va='bottom', fontsize=9)

plt.tight_layout()
plt.show()
```

#### % of Sales by Category

```python
category_sales = df.groupby("pizza_category")["total_price"].sum()

Category_pct = category_sales / category_sales.sum() * 100

plt.figure(figsize=(4,4))

colors = plt.get_cmap("tab20").colors

plt.pie(Category_pct, labels=Category_pct.index, colors=colors, autopct='%1.1f%%', wedgeprops=dict(width=0.4))

plt.title("Percentage of Sales by Pizza Category")
plt.show()
```

#### % of Sales by pizza size & Category

```python
sales_pivot = df.pivot_table(index="pizza_category", columns="pizza_size", values="total_price", aggfunc="sum", fill_value=0)

sales_pct = sales_pivot / sales_pivot.sum().sum() * 100

plt.figure(figsize=(6, 4))

sns.heatmap(sales_pct, annot=True, cmap="coolwarm", fmt=".1f")

plt.title("% of Sales by Pizza Size & Category")
plt.tight_layout()
plt.show()
```

#### Total Pizza Sold by pizza Category

```python
pizza_by_category = df.groupby("pizza_category")["quantity"].sum()

colors = list(plt.get_cmap("tab20").colors)
colors = colors[:len(pizza_by_category)]

ax = pizza_by_category.plot(kind="bar", figsize=(8, 5), color=colors, edgecolor="black")

ax.set_title("Total Pizza Sold by Pizza Category")
ax.set_xlabel("Category")
ax.set_ylabel("Quantity")

ax.bar_label(ax.containers[0], fmt="%d")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

#### Top 5 Best Selling Pizzas - Quantity
